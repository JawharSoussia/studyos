```python
from dotenv import load_dotenv
import os

load_dotenv()

openai_key = os.getenv("OPENAI_API_KEY")
mistral_key = os.getenv("MISTRAL_API_KEY")

assert openai_key, "OPENAI_API_KEY missing — check .env"
assert mistral_key, "MISTRAL_API_KEY missing — check .env"

print("Hello StudyOS")
print("OpenAI key loaded:", openai_key[:7] + "...")
print("Mistral key loaded:", mistral_key[:7] + "...")
```
